```python
import os
import glob
import shutil
import xml.etree.ElementTree as ET
from pathlib import Path
from sklearn.model_selection import train_test_split
from ultralytics import YOLO

# 1. Dataset Root Path
DATASET_ROOT = Path(r"C:\Users\user\Documents\GitHub\ImageProcessingPCB\Assignment\dataset\PCB_DATASET")
OUTPUT_DIR = Path(r"C:\Users\user\Documents\GitHub\ImageProcessingPCB\pcb_yolo_dataset")

# Clean previous failed dataset build if it exists
if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)

for folder in ["images/train", "images/val", "labels/train", "labels/val"]:
    os.makedirs(OUTPUT_DIR / folder, exist_ok=True)

# Defect class mapping
CLASSES = [
    "missing_hole",
    "mouse_bite",
    "open_circuit",
    "short",
    "spur",
    "spurious_copper"
]
# Normalize keys (lowercase without spaces or underscores) for flexible matching
class_to_id = {c.lower().replace("_", "").replace(" ", ""): i for i, c in enumerate(CLASSES)}

# 2. Build XML Map (Finds XML annotations anywhere in your directory)
xml_files = glob.glob(str(DATASET_ROOT / "**" / "*.xml"), recursive=True)
xml_map = {Path(f).stem: f for f in xml_files}

print(f"Found {len(xml_files)} XML annotation files.")

def convert_annotation(xml_path, output_txt_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    size = root.find('size')
    w = int(size.find('width').text)
    h = int(size.find('height').text)
    
    lines = []
    for obj in root.iter('object'):
        raw_name = obj.find('name').text.strip().lower().replace("_", "").replace(" ", "")
        if raw_name not in class_to_id:
            continue
        cls_id = class_to_id[raw_name]
        
        xmlbox = obj.find('bndbox')
        xmin = float(xmlbox.find('xmin').text)
        xmax = float(xmlbox.find('xmax').text)
        ymin = float(xmlbox.find('ymin').text)
        ymax = float(xmlbox.find('ymax').text)
        
        # YOLO format normalized values
        x_center = ((xmin + xmax) / 2.0) / w
        y_center = ((ymin + ymax) / 2.0) / h
        box_w = (xmax - xmin) / w
        box_h = (ymax - ymin) / h
        
        lines.append(f"{cls_id} {x_center:.6f} {y_center:.6f} {box_w:.6f} {box_h:.6f}")
        
    with open(output_txt_path, 'w') as f:
        f.write("\n".join(lines))
    return len(lines)

# 3. Gather Images and Split
all_images = glob.glob(str(DATASET_ROOT / "**" / "*.jpg"), recursive=True) + \
             glob.glob(str(DATASET_ROOT / "**" / "*.png"), recursive=True)

print(f"Found {len(all_images)} total PCB images.")

train_imgs, val_imgs = train_test_split(all_images, test_size=0.2, random_state=42)

def process_split(img_list, split_name):
    converted_count = 0
    for img_path in img_list:
        stem = Path(img_path).stem
        img_name = Path(img_path).name
        txt_name = f"{stem}.txt"
        
        # Copy Image
        shutil.copy(img_path, OUTPUT_DIR / "images" / split_name / img_name)
        
        # Convert XML to YOLO TXT Label
        if stem in xml_map:
            num_boxes = convert_annotation(xml_map[stem], OUTPUT_DIR / "labels" / split_name / txt_name)
            if num_boxes > 0:
                converted_count += 1
                
    print(f"[{split_name.upper()}] Successfully matched and converted {converted_count} annotated files.")

process_split(train_imgs, "train")
process_split(val_imgs, "val")

# 4. Generate YAML Configuration
yaml_content = f"""
path: {OUTPUT_DIR.resolve()}
train: images/train
val: images/val

names:
  0: missing_hole
  1: mouse_bite
  2: open_circuit
  3: short
  4: spur
  5: spurious_copper
"""

yaml_path = OUTPUT_DIR / "data.yaml"
with open(yaml_path, "w") as f:
    f.write(yaml_content.strip())

print("\n--- Setup Complete ---")

# 5. Start Training
model = YOLO('yolov8m.pt')
results = model.train(
    data=str(yaml_path.resolve()),
    epochs=50,
    imgsz=640,
    batch=16,
    name='pcb_defect_model',
    save=True
)
```

```python
def convert_annotation(xml_path, output_txt_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    size = root.find('size')
    w = int(size.find('width').text)
    h = int(size.find('height').text)
    
    lines = []
    for obj in root.iter('object'):
        cls_name = obj.find('name').text.strip().lower().replace(" ", "_")
        if cls_name not in class_to_id:
            continue
        cls_id = class_to_id[cls_name]
        
        xmlbox = obj.find('bndbox')
        xmin = float(xmlbox.find('xmin').text)
        xmax = float(xmlbox.find('xmax').text)
        ymin = float(xmlbox.find('ymin').text)
        ymax = float(xmlbox.find('ymax').text)
        
        # Convert bounding box coordinates to YOLO normalized center format
        x_center = ((xmin + xmax) / 2.0) / w
        y_center = ((ymin + ymax) / 2.0) / h
        box_w = (xmax - xmin) / w
        box_h = (ymax - ymin) / h
        
        lines.append(f"{cls_id} {x_center:.6f} {y_center:.6f} {box_w:.6f} {box_h:.6f}")
        
    with open(output_txt_path, 'w') as f:
        f.write("\n".join(lines))
```

```python
# Output folder where processed YOLO dataset will live
OUTPUT_DIR = Path("./pcb_yolo_dataset")
for folder in ["images/train", "images/val", "labels/train", "labels/val"]:
    os.makedirs(OUTPUT_DIR / folder, exist_ok=True)

# Find all images recursively inside your local dataset path
all_images = glob.glob(str(DATASET_ROOT / "**" / "*.jpg"), recursive=True)
print(f"Total images found in local path: {len(all_images)}")

# Split into 80% Training and 20% Validation
train_imgs, val_imgs = train_test_split(all_images, test_size=0.2, random_state=42)

def process_split(img_list, split_name):
    for img_path in img_list:
        xml_path = os.path.splitext(img_path)[0] + ".xml"
        img_name = os.path.basename(img_path)
        txt_name = os.path.splitext(img_name)[0] + ".txt"
        
        # Copy image file
        shutil.copy(img_path, OUTPUT_DIR / "images" / split_name / img_name)
        
        # Convert and write corresponding XML label to TXT
        if os.path.exists(xml_path):
            convert_annotation(xml_path, OUTPUT_DIR / "labels" / split_name / txt_name)

process_split(train_imgs, "train")
process_split(val_imgs, "val")

# Generate data.yaml with absolute Windows paths
yaml_content = f"""
path: {OUTPUT_DIR.resolve()}
train: images/train
val: images/val

names:
  0: missing_hole
  1: mouse_bite
  2: open_circuit
  3: short
  4: spur
  5: spurious_copper
"""

yaml_path = OUTPUT_DIR / "data.yaml"
with open(yaml_path, "w") as f:
    f.write(yaml_content.strip())

print(f"Dataset formatted! Configuration saved to: {yaml_path.resolve()}")
```

```python
from ultralytics import YOLO

# Load pre-trained model
model = YOLO('yolov8m.pt')

# Train model
results = model.train(
    data=str((OUTPUT_DIR / "data.yaml").resolve()),
    epochs=50,
    imgsz=640,
    batch=16,
    name='pcb_defect_model',
    save=True
)
```

```python
import cv2
import matplotlib.pyplot as plt

# Load trained weights
best_model = YOLO('runs/detect/pcb_defect_model/weights/best.pt')

# Test on a validation sample
val_images = glob.glob(str(OUTPUT_DIR / "images" / "val" / "*.jpg"))
if val_images:
    sample_img = val_images[0]
    
    # Run prediction
    results = best_model.predict(source=sample_img, conf=0.25)
    annotated_frame = results[0].plot()

    # Display image
    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title("PCB Defect Detection Test")
    plt.show()
```
