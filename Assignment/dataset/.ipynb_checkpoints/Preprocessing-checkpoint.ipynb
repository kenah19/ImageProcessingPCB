```python
import os
import glob
import random
import shutil
import cv2 as cv
import numpy as np

def adjust_brightness(image, beta=30):
    return cv.convertScaleAbs(image, alpha=1.0, beta=beta)

def adjust_exposure(image, gamma=1.2):
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in range(256)]).astype("uint8")
    return cv.LUT(image, table)

def add_gaussian_noise(image, mean=0, sigma=20):
    gauss = np.random.normal(mean, sigma, image.shape).astype('float32')
    noisy_image = np.clip(image.astype('float32') + gauss, 0, 255)
    return noisy_image.astype('uint8')

def augment_dataset(image_dir, annotation_dir, output_img_dir, output_xml_dir, sample_ratio=0.3):
    """
    Augments a random percentage of images and copies corresponding XML annotations.
    
    :param image_dir: Path to source image folder
    :param annotation_dir: Path to source XML folder
    :param output_img_dir: Path to save augmented images
    :param output_xml_dir: Path to save duplicated XML annotations
    :param sample_ratio: Percentage of files to augment (e.g., 0.3 for 30%)
    """
    os.makedirs(output_img_dir, exist_ok=True)
    os.makedirs(output_xml_dir, exist_ok=True)

    # Gather all image paths
    image_extensions = ('*.jpg', '*.jpeg', '*.png', '*.bmp')
    image_paths = []
    for ext in image_extensions:
        image_paths.extend(glob.glob(os.path.join(image_dir, ext)))

    # Calculate target count based on percentage
    total_images = len(image_paths)
    num_to_augment = int(total_images * sample_ratio)
    
    # Randomly select subset of files
    selected_images = random.sample(image_paths, num_to_augment)
    print(f"Selected {num_to_augment} out of {total_images} images ({sample_ratio * 100}%).")

    augmentation_types = ['bright', 'exposed', 'noisy']

    for img_path in selected_images:
        base_name = os.path.splitext(os.path.basename(img_path))[0]
        ext = os.path.splitext(img_path)[1]
        xml_path = os.path.join(annotation_dir, f"{base_name}.xml")

        img = cv.imread(img_path)
        if img is None:
            continue

        # Randomly choose one augmentation type per selected image
        aug_type = random.choice(augmentation_types)
        
        if aug_type == 'bright':
            aug_img = adjust_brightness(img, beta=random.randint(25, 45))
        elif aug_type == 'exposed':
            aug_img = adjust_exposure(img, gamma=random.uniform(1.3, 1.7))
        elif aug_type == 'noisy':
            aug_img = add_gaussian_noise(img, sigma=random.randint(15, 30))

        # Define new filenames
        new_base_name = f"{base_name}_{aug_type}"
        new_img_path = os.path.join(output_img_dir, f"{new_base_name}{ext}")
        new_xml_path = os.path.join(output_xml_dir, f"{new_base_name}.xml")

        # Save augmented image
        cv.imwrite(new_img_path, aug_img)

        # Copy original XML annotation with the updated file name
        if os.path.exists(xml_path):
            shutil.copy(xml_path, new_xml_path)

    print("Augmentation complete!")
```

```python
if __name__ == "__main__":
    augment_dataset(
        image_dir='~/Assignment/dataset/PCB_DATASET/images',
        annotation_dir='~/Assignment/dataset/PCB_DATASET/annotations',
        output_img_dir='~/Assignment/dataset/PCB_DATASET/augmented_images',
        output_xml_dir='~/Assignment/dataset/PCB_DATASET/augmented_annotations',
        sample_ratio=0.7  # Ratio of image to augment
    )
```
